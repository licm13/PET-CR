# 跷跷板游戏——神奇的互补关系
# The Seesaw Game - The Amazing Complementary Relationship

## 🤔 一个反直觉的现象 | A Counter-Intuitive Phenomenon

**中文**：想象一个奇怪的现象：当土壤变干、实际蒸发减少时，大气的"口渴程度"（潜在蒸发）反而增加！这就像天气在"嘲笑"干旱——越缺水，空气越想吸水。

**English**: Imagine a strange phenomenon: when soil becomes dry and actual evaporation decreases, the "thirstiness" of the atmosphere (potential evaporation) actually increases! It's like the weather is "mocking" drought - the more water-scarce, the more the air wants to absorb water.

这就是**互补关系（Complementary Relationship）**的核心思想！

This is the core idea of the **Complementary Relationship**!

---

## 🎢 跷跷板类比 | Seesaw Analogy

想象一个跷跷板，两端分别是：

Imagine a seesaw with two ends:

```
       实际蒸发 Ea ↑               潜在蒸发 Ep ↓
           (湿润条件)                (干旱条件)
           (Wet conditions)          (Dry conditions)
                    \              /
                     \            /
                      \          /
                       \        /
                        \      /
                         \    /
                          \  /
                           \/
                           /\
                          支点 (能量守恒)
                          Fulcrum (Energy conservation)
```

### 物理机制 | Physical Mechanism:

1. **湿润表面**（如湖泊）：
   - 实际蒸发 **Ea 很高**（水充足，尽情蒸发）
   - 潜在蒸发 **Ep 较低**（空气湿润，"不渴"）

   **Wet surface** (like a lake):
   - Actual evaporation **Ea is high** (plenty of water, evaporate freely)
   - Potential evaporation **Ep is low** (humid air, "not thirsty")

2. **干旱表面**（如沙漠）：
   - 实际蒸发 **Ea 很低**（水不足，无法蒸发）
   - 潜在蒸发 **Ep 很高**（空气干燥，"非常渴"）

   **Dry surface** (like a desert):
   - Actual evaporation **Ea is low** (insufficient water, cannot evaporate)
   - Potential evaporation **Ep is high** (dry air, "very thirsty")

**关键洞察 | Key Insight**: Ea 下降时，本应用于蒸发的能量转化为感热（加热空气），导致空气温度升高、湿度降低，从而 Ep 升高！

When Ea decreases, the energy that should have been used for evaporation is converted to sensible heat (heating the air), causing air temperature to rise and humidity to decrease, thus increasing Ep!

---

## 📐 Bouchet 假设（1963）| Bouchet Hypothesis (1963)

法国科学家 Bouchet 在 1963 年提出了一个简洁的数学表达：

French scientist Bouchet proposed a concise mathematical expression in 1963:

$$
E_a + E_p = 2 E_w
$$

其中 | Where:
- $E_a$: 实际蒸散发（Actual ET）
- $E_p$: 潜在蒸散发（Penman potential ET）
- $E_w$: 湿环境蒸散发（Wet-environment ET, Priestley-Taylor）

**物理意义 | Physical Meaning**:
- $E_w$ 是"能量上限"——假设水分无限供应时的蒸发
- $E_w$ is the "energy ceiling" - evaporation when water supply is unlimited
- 当水分减少时，$E_a$ 下降的量 ≈ $E_p$ 上升的量（互补）
- When water decreases, the amount $E_a$ decreases ≈ the amount $E_p$ increases (complementary)

## 🧪 实验 1: 对比不同湿度条件下的 CR 模型 | Experiment 1: Compare CR Models Under Different Moisture Conditions

````python
# 设置气象条件 | Set meteorological conditions
# 假设 Ep 和 Ew 已知（通常从气象数据计算得到）
# Assume Ep and Ew are known (usually calculated from meteorological data)

# Penman potential ET (W/m²)
ep = 400.0  

# Priestley-Taylor ET (W/m²) - 通常比 Ep 低，因为它假设湿润条件
# Usually lower than Ep because it assumes wet conditions
ew = 300.0  

print(f"输入条件 | Input conditions:")
print(f"  Penman potential ET (Ep) = {ep} W/m²")
print(f"  Priestley-Taylor ET (Ew) = {ew} W/m²")
print()

# 使用不同的 CR 模型估算实际 ET | Estimate actual ET using different CR models

# 1. Bouchet 模型（最简单）| Bouchet model (simplest)
ea_bouchet = petcr.bouchet_cr(ep, ew)
print(f"1️⃣ Bouchet CR:")
print(f"   实际 ET (Ea) = {ea_bouchet:.2f} W/m²")
print(f"   公式 | Formula: Ea = 2*Ew - Ep = 2*{ew} - {ep} = {ea_bouchet:.2f}")
print()

# 2. Sigmoid CR（Han & Tian 2018）- 更灵活
# Sigmoid CR (Han & Tian 2018) - more flexible
beta = 1.26  # 经验参数 | Empirical parameter
ea_sigmoid = petcr.sigmoid_cr(ep, ew, beta)
print(f"2️⃣ Sigmoid CR (β={beta}):")
print(f"   实际 ET (Ea) = {ea_sigmoid:.2f} W/m²")
print()

# 3. Polynomial CR（Brutsaert 2015）
ea_polynomial = petcr.polynomial_cr(ep, ew)
print(f"3️⃣ Polynomial CR:")
print(f"   实际 ET (Ea) = {ea_polynomial:.2f} W/m²")
print()

# 验证互补关系 | Verify complementary relationship
print("═" * 60)
print("🔍 验证互补关系 | Verify Complementary Relationship")
print("═" * 60)
print(f"Ep + Ea (Bouchet)     = {ep + ea_bouchet:.2f} W/m²")
print(f"2 * Ew                = {2 * ew:.2f} W/m²")
print(f"误差 | Error          = {abs((ep + ea_bouchet) - 2*ew):.2f} W/m²")
print()
print("💡 Bouchet 模型完美满足互补关系！")
print("💡 Bouchet model perfectly satisfies the complementary relationship!")
````

## 📊 可视化：跷跷板效应 | Visualization: Seesaw Effect

让我们改变水分条件（通过改变 Ep），观察 Ea 和 Ep 的反向变化：

Let's change moisture conditions (by varying Ep) and observe the inverse relationship between Ea and Ep:

````python
# 固定 Ew，改变 Ep（模拟从湿润到干旱）
# Fix Ew, vary Ep (simulate from wet to dry)
ew_fixed = 300.0  # W/m²
ep_range = np.linspace(100, 500, 50)  # Ep 从 100 到 500 W/m²

# 计算不同模型的 Ea | Calculate Ea for different models
ea_bouchet_range = [petcr.bouchet_cr(ep_val, ew_fixed) for ep_val in ep_range]
ea_sigmoid_range = [petcr.sigmoid_cr(ep_val, ew_fixed, beta=1.26) for ep_val in ep_range]

# 绘图 | Plot
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# 左图：Ea vs Ep | Left: Ea vs Ep
ax1 = axes[0]
ax1.plot(ep_range, ea_bouchet_range, 'b-o', linewidth=2.5, markersize=5, 
         label='Bouchet CR', alpha=0.8)
ax1.plot(ep_range, ea_sigmoid_range, 'r--^', linewidth=2.5, markersize=5, 
         label='Sigmoid CR', alpha=0.8)
ax1.axhline(y=ew_fixed, color='green', linestyle=':', linewidth=2, 
            label=f'Ew = {ew_fixed} W/m² (能量上限 | Energy ceiling)')
ax1.axhline(y=0, color='gray', linestyle='-', linewidth=0.5)

ax1.set_xlabel('潜在蒸散发 Ep (W/m²) | Potential ET Ep (W/m²)', fontweight='bold', fontsize=12)
ax1.set_ylabel('实际蒸散发 Ea (W/m²) | Actual ET Ea (W/m²)', fontweight='bold', fontsize=12)
ax1.set_title('互补关系：Ea 随 Ep 的变化\nComplementary Relationship: Ea vs Ep', 
              fontweight='bold', fontsize=13)
ax1.legend(fontsize=11, loc='best')
ax1.grid(alpha=0.3)

# 添加注释 | Add annotations
ax1.annotate('湿润 → 干旱\nWet → Dry', 
             xy=(450, 100), fontsize=12, fontweight='bold',
             bbox=dict(boxstyle='round,pad=0.5', facecolor='yellow', alpha=0.5),
             ha='center')
ax1.annotate('Ea 下降\nEa decreases', 
             xy=(400, 180), xytext=(350, 250),
             arrowprops=dict(arrowstyle='->', lw=2, color='red'),
             fontsize=11, fontweight='bold', color='red')

# 右图：互补性验证（Ea + Ep vs 2*Ew）
# Right: Complementarity verification (Ea + Ep vs 2*Ew)
ax2 = axes[1]
sum_bouchet = np.array(ea_bouchet_range) + ep_range
sum_sigmoid = np.array(ea_sigmoid_range) + ep_range
target_2ew = np.full_like(ep_range, 2 * ew_fixed)

ax2.plot(ep_range, sum_bouchet, 'b-o', linewidth=2.5, markersize=5, 
         label='Ea + Ep (Bouchet)', alpha=0.8)
ax2.plot(ep_range, sum_sigmoid, 'r--^', linewidth=2.5, markersize=5, 
         label='Ea + Ep (Sigmoid)', alpha=0.8)
ax2.axhline(y=2*ew_fixed, color='green', linestyle=':', linewidth=3, 
            label=f'2 * Ew = {2*ew_fixed} W/m²')

ax2.set_xlabel('潜在蒸散发 Ep (W/m²) | Potential ET Ep (W/m²)', fontweight='bold', fontsize=12)
ax2.set_ylabel('Ea + Ep (W/m²)', fontweight='bold', fontsize=12)
ax2.set_title('互补关系验证：Ea + Ep = 2*Ew\nComplementarity Verification', 
              fontweight='bold', fontsize=13)
ax2.legend(fontsize=11, loc='best')
ax2.grid(alpha=0.3)

# 添加文字说明 | Add text explanation
ax2.text(0.5, 0.05, 
         'Bouchet 模型完美满足互补关系（直线）\n'
         'Sigmoid 模型略有偏差（更接近实际观测）\n'
         'Bouchet model perfectly satisfies CR (straight line)\n'
         'Sigmoid model has slight deviation (closer to observations)',
         transform=ax2.transAxes,
         fontsize=10,
         verticalalignment='bottom',
         horizontalalignment='center',
         bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.savefig('figures/complementary_relationship.png', dpi=150, bbox_inches='tight')
plt.show()

print("\n✅ 可视化完成！Visualization complete!")
````

## 🌍 真实案例：干旱的反直觉现象 | Real Case: Counter-Intuitive Drought Phenomenon

### 场景设定 | Scenario Setting:

假设一个地区从**正常年份**变为**干旱年份**：

Suppose a region changes from **normal year** to **drought year**:

- **正常年份 | Normal year**: 土壤湿润，植被茂盛
- **干旱年份 | Drought year**: 降雨减少50%，土壤干燥

````python
# 定义两种场景 | Define two scenarios
scenarios = {
    '正常年份\nNormal': {'ep': 350, 'ew': 280, 'color': 'blue'},
    '干旱年份\nDrought': {'ep': 450, 'ew': 250, 'color': 'red'}
}

results = {}
for name, params in scenarios.items():
    ep_val = params['ep']
    ew_val = params['ew']
    ea_val = petcr.bouchet_cr(ep_val, ew_val)
    results[name] = {'ep': ep_val, 'ew': ew_val, 'ea': ea_val, 'color': params['color']}

# 绘制对比图 | Plot comparison
fig, ax = plt.subplots(figsize=(12, 7))

x_pos = np.arange(len(scenarios))
width = 0.25

# 绘制柱状图 | Plot bar chart
for i, (name, data) in enumerate(results.items()):
    color = data['color']
    alpha_vals = [0.4, 0.7, 1.0]
    
    ax.bar(i - width, data['ep'], width, label='Ep' if i == 0 else '', 
           color=color, alpha=alpha_vals[0], edgecolor='black', linewidth=1.5)
    ax.bar(i, data['ew'], width, label='Ew' if i == 0 else '', 
           color=color, alpha=alpha_vals[1], edgecolor='black', linewidth=1.5)
    ax.bar(i + width, data['ea'], width, label='Ea' if i == 0 else '', 
           color=color, alpha=alpha_vals[2], edgecolor='black', linewidth=1.5)
    
    # 添加数值标签 | Add value labels
    ax.text(i - width, data['ep'] + 10, f"{data['ep']:.0f}", ha='center', fontweight='bold')
    ax.text(i, data['ew'] + 10, f"{data['ew']:.0f}", ha='center', fontweight='bold')
    ax.text(i + width, data['ea'] + 10, f"{data['ea']:.0f}", ha='center', fontweight='bold')

ax.set_ylabel('蒸散发 (W/m²) | Evapotranspiration (W/m²)', fontweight='bold', fontsize=13)
ax.set_title('正常年份 vs 干旱年份的蒸散发对比\nNormal vs Drought Year ET Comparison', 
             fontweight='bold', fontsize=14)
ax.set_xticks(x_pos)
ax.set_xticklabels(list(scenarios.keys()), fontsize=12, fontweight='bold')
ax.legend(fontsize=12, loc='upper left')
ax.grid(axis='y', alpha=0.3)

# 添加关键发现 | Add key findings
ea_normal = results['正常年份\nNormal']['ea']
ea_drought = results['干旱年份\nDrought']['ea']
ep_normal = results['正常年份\nNormal']['ep']
ep_drought = results['干旱年份\nDrought']['ep']

change_ea = ((ea_drought - ea_normal) / ea_normal) * 100
change_ep = ((ep_drought - ep_normal) / ep_normal) * 100

finding_text = f"""
🔍 关键发现 | Key Findings:

干旱时 | During drought:
• 实际蒸发 Ea 下降 {abs(change_ea):.1f}% ⬇️
  (水分不足，蒸发受限)
  
• 潜在蒸发 Ep 上升 {change_ep:.1f}% ⬆️
  (空气更干燥，"更渴")
  
这就是互补关系！
This is the complementary relationship!
"""

ax.text(0.98, 0.97, finding_text,
        transform=ax.transAxes,
        fontsize=11,
        verticalalignment='top',
        horizontalalignment='right',
        bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8),
        family='monospace')

plt.tight_layout()
plt.savefig('figures/drought_case_study.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"\n实际蒸发变化 | Actual ET change: {change_ea:.1f}%")
print(f"潜在蒸发变化 | Potential ET change: {change_ep:.1f}%")
````

## 🎯 核心要点总结 | Key Takeaways

### 1. 互补关系的物理本质 | Physical Nature of CR

**能量守恒视角 | Energy Conservation Perspective**:

$$
R_n = LH + SH + G
$$

其中 | Where:
- $R_n$: 净辐射（总能量输入）| Net radiation (total energy input)
- $LH$: 潜热通量（蒸发消耗的能量）| Latent heat flux (energy for evaporation)
- $SH$: 感热通量（加热空气的能量）| Sensible heat flux (energy for heating air)
- $G$: 土壤热通量（存储在土壤中的能量）| Ground heat flux (energy stored in soil)

**关键机制 | Key Mechanism**:
1. 当水分减少 → LH ↓（蒸发减少）
2. 能量守恒 → SH ↑（更多能量加热空气）
3. 空气温度 ↑、湿度 ↓ → Ep ↑（大气需求增加）

When water decreases:
1. LH ↓ (less evaporation)
2. By energy conservation → SH ↑ (more energy heats air)
3. Air temperature ↑, humidity ↓ → Ep ↑ (atmospheric demand increases)
